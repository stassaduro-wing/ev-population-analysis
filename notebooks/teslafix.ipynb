```python
with open('Electric_Vehicle_Population_Data_changed(in).csv', 'rb') as f:
    data=f.read()
position = [i for i, b in enumerate(data) if b>127]
print(len(position))

for p in position[:10]:
    print(data[p-15:p+15])

pattern = b"TES??\x8bLA"
print('Pattern matches: ', data.count(pattern))
print('Matches total corrupt byte count?', data.count(pattern)==len(position))
```

```python
fixed = data.replace(pattern, b'TESLA')

fixed.decode('utf-8', errors='strict')
print('All is ok')

with open('Electric_Vehicle_Population_Data_changed(in).csv', 'wb') as f:
    f.write(fixed)

with open('Electric_Vehicle_Population_Data_changed(in).csv', 'r') as f:
    for i, line in enumerate(f):
        if 'TESLA' in line:
            print(line)
            break
```
